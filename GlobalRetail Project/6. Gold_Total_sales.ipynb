```sql
CREATE OR REPLACE TABLE globalretail.gr_gold.daily_sales
USING DELTA
AS
SELECT
    transaction_date,
   ROUND(SUM(total_amount),3) AS total_daily_sales
FROM globalretail.gr_silver.transaction_transform
GROUP BY transaction_date
ORDER BY transaction_date;
```

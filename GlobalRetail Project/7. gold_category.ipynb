```sql
use globalretail.gr_gold
```

```sql
create or replace table globalretail.gr_gold.daily_sales
as
select
p.category as product_category,
round(sum(o.total_amount),3)as category_total_sales
from
globalretail.gr_silver.transaction_transform o
join
globalretail.gr_silver.product_transform p
on
o.product_id = p.product_id
group by p.category
```

```sql
select * from globalretail.gr_gold.daily_sales
```

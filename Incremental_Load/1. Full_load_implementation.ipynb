```python
spark.conf.set("fs.azure.account.key.databricksrg2026.dfs.core.windows.net",account_key)
```

**Ingest circuits file**

```python
circuits_df = spark.read\
    .option("header", True)\
    .option("inferSchema", True)\
        .csv("abfss://raw@databricksrg2026.dfs.core.windows.net/2021-03-21/circuits.csv")
```

**Rename specific columns**

```python
circuits_rename_df = circuits_df.withColumnRenamed("circuitId", "circuit_id")\
    .withColumnRenamed("circuitRef", "circuit_ref")\
    .withColumnRenamed("lat", "latitude")\
    .withColumnRenamed("lng", "longitude")\
    .withColumnRenamed("alt", "altitude")\
.withColumnRenamed("name" , "circuits_name")\
  
```

```python
circuits_final_df = circuits_rename_df.write.mode("overwrite").parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/2021-03-21/circuits")
```

**Ingest Races file**

```python
races_df = spark.read\
    .option("header", True)\
    .option("inferSchema", True)\
        .csv("abfss://raw@databricksrg2026.dfs.core.windows.net/2021-03-21/races.csv")
```

```python
races_rename_df = races_df.withColumnRenamed("raceId", "race_id")\
    .withColumnRenamed("year", "race_year")\
        .withColumnRenamed("time", "race_time")\
            .withColumnRenamed("name", "race_name")\
                .withColumnRenamed("url", "race_url")

    
```

```python
races_final_df = races_rename_df.write.mode("overwrite").parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/2021-03-21/races")
```

**Ingest Drivers file**

```python
drivers_df = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
        .csv("abfss://raw@databricksrg2026.dfs.core.windows.net/2021-03-21/drivers.json")
```

```python
from pyspark.sql.functions import lit, concat,col
```

```python
driver_rename_df = drivers_df.withColumnRenamed("driverId","driver_id") \
                            .withColumnRenamed("driverRef","driver_ref") \
                            .withColumnRenamed("nationality", "driver_nationality")\
                            .withColumnRenamed("dob", "driver_dob")\
                            .withColumnRenamed("name","driver_name")
                                
```

```python
drivers_final_df = driver_rename_df.write.mode("overwrite").parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/2021-03-21/drivers")
```

1. Read the file

```python
account_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-strg-access-key' )
```

```python
spark.conf.set("fs.azure.account.key.databricksrg2026.dfs.core.windows.net",account_key)
```

define schema using ddl instead of structtype

```python
constructor_schema  =  "constructorId INT, constructorRef STRING, name STRING, nationality STRING,URL STRING"
```

```python
constructor_df = spark.read\
    .schema(constructor_schema)\
    .json("abfss://raw@databricksrg2026.dfs.core.windows.net/constructors.json")
```

```python
constructor_df.printSchema()
```

```python
display(constructor_df)
```

drop url column

```python
constructor_df.drop("URL")
```

Rename and add ingestion_date

```python
from pyspark.sql.functions import current_timestamp
```

```python
constructor_rename_df = constructor_df.withColumnRenamed("constructorId","constructor_id")\
    .withColumnRenamed("constructorRef","constructor_ref")\
        .withColumn("ingestion_date" , current_timestamp())


  
```

```python
display(constructor_rename_df)
```

write in parquet

add ingestion_date

```python
constructor_final_df = constructor_rename_df.write.mode("overwrite").parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/constructor")
```

```python
account_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-strg-access-key' )
```

```python
spark.conf.set("fs.azure.account.key.databricksrg2026.dfs.core.windows.net",account_key)
```

```python
pitstop_schema  = "raceId integer, driverId integer, stop string, lap integer, time string, duration string, milliseconds integer"
```

```python
pitstop_df = spark.read.json("abfss://raw@databricksrg2026.dfs.core.windows.net/pit_stops.json")
```

```python
from pyspark.sql.functions import current_date
```

```python
pitstop_transform_df = pitstop_df.withColumnRenamed("raceId","race_id") .withColumnRenamed("driverId","driver_id") .withColumn("ingestion_date",current_date())

```

```python
pitstop_transform_df.show()
```

```python
pitstop_transform_df.write.mode("overwrite").parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/pitstop")
```

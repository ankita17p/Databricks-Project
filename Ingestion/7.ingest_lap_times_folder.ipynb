```python
account_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-strg-access-key' )
```

```python
spark.conf.set("fs.azure.account.key.databricksrg2026.dfs.core.windows.net",account_key)
```

```python
lap_time_schema  = "raceId integer, driverId integer, lap integer, position integer, time string, milliseconds integer"
```

```python
lap_time_df =  spark.read.schema(lap_time_schema).csv("abfss://raw@databricksrg2026.dfs.core.windows.net/lap_times")
```

```python
display(lap_time_df)
```

```python
from pyspark.sql.functions import current_timestamp
```

```python
lap_time_rename  = lap_time_df.withColumnRenamed("raceId","race_id") \
                              .withColumnRenamed("driverId","driver_id") \
                              .withColumn("ingestion_date",current_timestamp())

```

```python
display(lap_time_rename)
```

```python
lap_time_rename.write.mode("overwrite").parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/lap_times")
```

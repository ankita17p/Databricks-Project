```python
df = spark.read.table("bronze.default.big_mart_sales")
```

```python
df.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number
```

```python

df.withColumn("Row_Number", row_number().over(Window.partitionBy("Item_Identifier").orderBy("Item_identifier"))).display()
```

```python
from pyspark.sql.functions import rank
```

```python
df.withColumn("rank",rank().over(Window.partitionBy("Item_Identifier").orderBy("Item_Weight"))).display()
```

```python
from pyspark.sql.functions import sum
```

```python
df.withColumn("cumsum",sum("Item_MRP").over(Window.orderBy("Item_Type").rowsBetween(Window.unboundedPreceding,Window.currentRow))).display()
```

```python
df.write.csv.save("dbfs:/FileStore/tables/bronze/data.csv")
```

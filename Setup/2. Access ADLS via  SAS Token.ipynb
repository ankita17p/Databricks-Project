1.Setup spark configuration
2.List files from demo container
3.Read data from circuit.csv files

```python
account_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-strg-access-key' )
```

```python
spark.conf.set("fs.azure.account.auth.type.databricksrg2026.dfs.core.windows.net", "SAS")
spark.conf.set("fs.azure.sas.token.provider.type.databricksrg2026.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.sas.FixedSASTokenProvider")
spark.conf.set("fs.azure.sas.fixed.token.databricksrg2026.dfs.core.windows.net", account_key)
```

```python
dbutils.fs.ls("abfss://demo@databricksrg2026.dfs.core.windows.net")
```

```python
display(dbutils.fs.ls("abfss://demo@databricksrg2026.dfs.core.windows.net/circuits.csv"))
```

```python
spark.read.csv("abfss://demo@databricksrg2026.dfs.core.windows.net/circuits.csv")
```

#Access Azure Data Lake using Service Principal
Steps to follow
--Register Azure AD Application / Service Principal
--Generate a secret/ password for the Application
-Set Spark Config with App/ Client Id, Directory/ Tenant Id & Secret
-Assign Role 'Storage Blob Data Contributor' to the Data Lake.

```python
client_id_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-sp-client-id' )
client_secret_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-sp-client-secret' )
tenant_id_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-sp-tenant-id' )
```

```python
client_id  = client_id_key
tenant_id = tenant_id_key
```

```python
Client_secret  = client_secret_key
```

```python
spark.conf.set("fs.azure.account.auth.type.databricksrg2026.dfs.core.windows.net", "OAuth")
spark.conf.set("fs.azure.account.oauth.provider.type.databricksrg2026.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
spark.conf.set("fs.azure.account.oauth2.client.id.databricksrg2026.dfs.core.windows.net", client_id)
spark.conf.set("fs.azure.account.oauth2.client.secret.databricksrg2026.dfs.core.windows.net", Client_secret)
spark.conf.set("fs.azure.account.oauth2.client.endpoint.databricksrg2026.dfs.core.windows.net", f"https://login.microsoftonline.com/{tenant_id}/oauth2/token")
```

```python
dbutils.fs.ls("abfss://demo@databricksrg2026.dfs.core.windows.net")
```

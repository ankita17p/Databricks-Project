```python
account_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-strg-access-key' )
```

```python
spark.conf.set("fs.azure.account.key.databricksrg2026.dfs.core.windows.net",account_key)
```

```python
circuit_df = spark.read.csv(
    "abfss://raw@databricksrg2026.dfs.core.windows.net/circuits.csv",
    header=True,
    inferSchema=True
)\
    .filter("circuitId < 70")\
        .withColumnRenamed("name","circuit_name")
races_df = spark.read.parquet("abfss://processed@databricksrg2026.dfs.core.windows.net/races",
                          header=True,
                          inferSchema=True)\
                              .withColumnRenamed("name","race_name")
```

```python
races_filter_df =  races_df.filter("race_year = 2019")
```

```python
display(races_filter_df)
display(circuit_df)
```

apply join on circuitid

**Inner Join**

```python
race_circuit_df =  circuit_df.join(races_filter_df , circuit_df.circuitId == races_filter_df.circuit_id,"inner")\
  .select(circuit_df.name ,circuit_df.location,circuit_df.country,races_filter_df.race_name,races_filter_df.race_year,races_filter_df.round)
```

```python
display(race_circuit_df)
```

**Outer Join**

Circuit which don't have races

```python
race_circuit_outer_df =  circuit_df.join(races_filter_df , circuit_df.circuitId == races_filter_df.circuit_id,"left")\
  .select(circuit_df.circuit_name ,circuit_df.location,circuit_df.country,races_filter_df.race_name,races_filter_df.race_year,races_filter_df.round)
```

```python
display(race_circuit_outer_df)
```

**Right outer join**

```python
race_circuit_right_df =  circuit_df.join(races_filter_df , circuit_df.circuitId == races_filter_df.circuit_id,"right")\
  .select(circuit_df.circuit_name ,circuit_df.location,circuit_df.country,races_filter_df.race_name,races_filter_df.race_year,races_filter_df.round)
```

```python
display(race_circuit_right_df)
```

```python
race_circuit_full_df =  circuit_df.join(races_filter_df , circuit_df.circuitId == races_filter_df.circuit_id,"full")\
  .select(circuit_df.circuit_name ,circuit_df.location,circuit_df.country,races_filter_df.race_name,races_filter_df.race_year,races_filter_df.round)
```

```python
display(race_circuit_full_df)
```

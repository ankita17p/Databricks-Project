```python
%sql
drop table if exists formula1_dev.bronze.drivers;
create table if not exists formula1_dev.bronze.drivers
(
  driverId int,
  driverRef string,
  number int,
  code string,
  name struct<forename:string, surname:string>,
  nationality string,
  dob date,
  url string
)
using json
options(path "abfss://bronze@databricksucproj26.dfs.core.windows.net/drivers.json");
```

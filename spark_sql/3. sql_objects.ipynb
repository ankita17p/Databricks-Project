**View in Table**
- Create temp View
- Global Temp View
- Permanent view

```sql
drop view v_race_results
```

```sql
create temp view v_race_results
as
select *
from demo.race_results_python where race_year = '2020'
```

```sql
SELECT * FROM v_race_results
```

**Global Temp View**

```sql
create or replace  global temp view gv_race_results
as
select *
from demo.race_results_python where race_year = '2012'
```

```sql
select * from global_temp.gv_race_results
```

```sql
show tables
```

```sql
show tables in global_temp
```

**Permanent View**

```sql
create or replace  view pv_race_results
as
select *
from demo.race_results_python where race_year = '2000'
```

```sql
%python
account_key = dbutils.secrets.get(scope='databricks-scope' ,key = 'databricks-strg-access-key' )
```

```sql
%python
spark.conf.set("fs.azure.account.key.databricksrg2026.dfs.core.windows.net",account_key)
```

```sql
%python
race_results_df = spark.read.parquet("abfss://presentation@databricksrg2026.dfs.core.windows.net/")
```

```sql
SELECT current_catalog();
```

```sql
SHOW CATALOGS
```

```sql
USE CATALOG databricks_proj
```

```sql
%python
race_results_df.write.format("delta").mode("overwrite").saveAsTable("demo.race_results_python")
```

```sql
use demo;
SHOW tables;
```

```sql
DESCRIBE EXTENDED race_results_python
```

```sql
select * from demo.race_results_python
where race_year = '2020'
```

```sql
create table race_result_sql
as
select * from demo.race_results_python
where race_year = '2020'
```

```sql
describe extended race_result_sql
```

```sql
drop table if exists race_result_sql
```
